```python
import pandas as pd
```

```python
from pathlib import Path
import json

from recsys_large_scale.utils.config import load_config

# works whether the kernel's cwd is the notebook's own folder or the project root
project_root = Path.cwd() if (Path.cwd() / "configs").exists() else Path.cwd().parent

data_config = load_config(project_root / "configs" / "data_movielens25m.yaml")
processed_dir = project_root / data_config["processed_dir"]

id_maps = json.loads((processed_dir / "id_maps.json").read_text())
print(f"n_users={id_maps['n_users']:,}  n_items={id_maps['n_items']:,}")

parquet_files = sorted(processed_dir.glob("*.parquet"))
print(f"\nFound {len(parquet_files)} parquet file(s): {[p.name for p in parquet_files]}")

for path in parquet_files:
    df = pd.read_parquet(path)
    print(f"\n{'=' * 70}\n{path.name}\n{'=' * 70}")
    print(f"shape: {df.shape}")
    print(f"dtypes:\n{df.dtypes}")
    print("\n3 random rows:")
    print(df.sample(3, random_state=42))
    print("\nsummary stats:")
    print(df.describe())
```
